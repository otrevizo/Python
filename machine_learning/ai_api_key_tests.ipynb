```python
import os

for key_name in ["ANTHROPIC_API_KEY", "OPENAI_API_KEY"]:
    value = os.environ.get(key_name)
    if value:
        print(f"{key_name}: SET (starts with {value[:6]}..., length {len(value)})")
    else:
        print(f"{key_name}: NOT SET")
```
